```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
```

```python
qc = QuantumCircuit(1, 1) # Create a quantum circuit with 1 qubit and 1 classical bit to store the measurement result


qc.h(0) #Apply the Hadamard gate to create superposition (the "coin flip in mid-air")


qc.measure(0, 0)  #Measure the qubit forcing the coin to land on either 0 or 1)


#The circuit diagram
display(qc.draw('mpl'))
```

```python
# Initialize the simulator
simulator = AerSimulator()

# Run the circuit 1,000 times (shots)
job = simulator.run(qc, shots=1000)
result = job.result()
counts = result.get_counts(qc)

# Display the raw counts
print("Flip Results (0 = Heads, 1 = Tails):", counts)
```

Plotting a histogram showing the frequency of Heads vs. Tails. Because of quantum superposition, The bars should be roughly 50/50

```python
# Plot a visual bar chart of our quantum coin flips
fig = plot_histogram(counts, title="Quantum Coin Toss Results (1,000 Flips)")
display(fig)
```
